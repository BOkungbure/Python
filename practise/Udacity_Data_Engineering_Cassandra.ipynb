```python
import cassandra as cass
```

```python
from cassandra.cluster import Cluster
try:
    cluster = Cluster(['127.0.0.1'])
    session = cluster.connect()
    print("Cassandra connection established successfully.")
except Exception as e:
    print(f"Failed to connect to Cassandra: {e}")
```

```python
try:
    session.execute("""select * from music_library""")
except Exception as e:
    print(f"Query failed: {e}")
```

```python
try:
    session.execute("""
    CREATE KEYSPACE IF NOT EXISTS udacity
    WITH REPLICATION = 
    { 'class' : 'SimpleStrategy', 'replication_factor' : 1 }""")
except Exception as e:
    print(f"Keyspace creation failed: {e}")
```

```python
try:
    session.set_keyspace('udacity')
    print("Keyspace set to 'udacity'.")
except Exception as e:
    print(f"Setting keyspace failed: {e}")
```

```python
query = "CREATE TABLE IF NOT EXISTS music_library"
query = query + "(year int, artist_name text, album_name text, PRIMARY KEY (year, artist_name))"

try:
    session.execute(query)
    print("Table 'music_library' created successfully.")
except Exception as e:
    print(f"Table creation failed: {e}")
```

```python
query = "select count(*) from music_library"

try:
    count = session.execute(query)
except Exception as e:
    print(e)

print(count.one())
```

```python
query = "INSERT INTO music_library (year, artist_name, album_name)"
query = query + " VALUES (%s, %s, %s)  "

try:
    session.execute(query, (1970, 'The Beatles', 'Let It Be'))
    session.execute(query, (1965, 'The Beatles', 'Rubber Soul'))
    session.execute(query, (1966, 'The Beatles', 'Revolver'))
    print("Data inserted successfully.")
except Exception as e:
    print(f"Data insertion failed: {e}")
```

```python
query = "SELECT * FROM music_library"

try:
    rows = session.execute(query)
except Exception as e:
    print(f"Data retrieval failed: {e}")

for row in rows:
    print(f"{row.year} {row.artist_name} {row.album_name}")
```

```python
query = "drop table music_library"

try:
    session.execute(query)
    print("Table 'music_library' dropped successfully.")
except Exception as e:
    print(f"Failed to drop table 'music_library': {e}")
```

```python
session.shutdown()
cluster.shutdown()
```
