```python
import psycopg2 as psql
import keyring as kr
```

```python
SERVICE_NAME  = 'PostGres'
USERNAME = 'postgres'
PASSWORD = kr.get_password(SERVICE_NAME, USERNAME)
HOST = 'localhost'
PORT = 5432


try:
    print("...1. Attempting to connect to the database. ⌛...")
    conn = psql.connect(
        host = HOST,
        dbname = 'studentdb',
        user = USERNAME,
        password = PASSWORD
    )
except psql.Error as e:
    print("===!Attempted to connect to the database but failed!===")
    print(e)
else:
    print("...2. Connection to the database was successful. ✅...")
finally:
    """ Nested try-except-finally block to create a cursor object. """
    try:
        cur = conn.cursor()
    except psql.Error as e:
        print("Could not create a cursor object.")
        print(e)
    else:
        print("...3. Cursor object created successfully. ✅...")
    print("End of database connection attempt.")
```

```python
conn.set_session(autocommit=True)
```

```python
try:
    print("...4. Attempting to create the album_data table. ⌛...")
    cur.execute("CREATE TABLE IF NOT EXISTS album_data (album_name varchar, artist_name varchar, release_year int, genre varchar);")
except psql.Error as e:
    print("===!Attempted to create the album_data table but failed!===")
    print(e)
else:
    print("...5. album_data table created successfully. ✅...")
```

```python
try:
    cur.execute("INSERT INTO album_data (album_name, artist_name, release_year, genre) VALUES (%s, %s, %s, %s);",
                ('The Dark Side of the Moon', 'Pink Floyd', 1973, 'Progressive Rock'))
except psql.Error as e:
    print("=== !Attempted to insert data into album_data table but failed! ===")
    print(e)
else:
    print("...6. Data inserted into album_data table successfully. ✅...")
```

```python
try:
    cur.execute("INSERT INTO album_data (album_name, artist_name, release_year, genre) VALUES (%s, %s, %s, %s);",
                ('Let it Be', 'The Beatles', 1970, 'Rock'))
except psql.Error as e:
    print("=== !Attempted to insert data into album_data table but failed! ===")
    print(e)
else:
    print("...7. Data inserted into album_data table successfully. ✅...")
```

```python
try:
    cur.execute("SELECT * FROM album_data;")
except psql.Error as e:
    print("=== !Attempted to retrieve data from album_data table but failed! ===")
    print(e)

row = cur.fetchone()
while row:
    print(row)
    row = cur.fetchone()
```

```python
cur.close()
conn.close()
```
